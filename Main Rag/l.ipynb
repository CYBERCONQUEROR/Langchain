```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled
from langchain_community.document_loaders import YoutubeLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python
video_url = "https://www.youtube.com/watch?v=f9Aje_cN_CY&t=8102s"

try:
    # YoutubeLoader is specifically designed for LangChain and handles 
    # transcript selection and parsing much more reliably.
    loader = YoutubeLoader.from_youtube_url(
        video_url, 
        add_video_info=False, 
        language=["hi"] # Forces Hindi transcript
    )
    docs = loader.load()
    
    if docs:
        transcript = docs[0].page_content
        print("Transcript loaded successfully!")
        print(transcript[:500] + "...") # Preview first 500 characters
    else:
        print("No transcript found for the requested language.")
        transcript = ""

except Exception as e:
    print(f"Failed to fetch transcript: {e}")
    transcript = ""
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 4})
```

```python
retriever
```

```python
retriever.invoke("what is claude")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(
    model="gemini-3.1-pro-preview",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
prompt = PromptTemplate(
    template="""
      You are a helpful assistant.
      Answer ONLY from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables = ['context', 'question']
)
```

```python
question          = "is there calude code get discussed"
retrieved_docs    = retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
context_text
```

```python
final_prompt = prompt.invoke({"context": context_text, "question": question})
```

```python
final_prompt
```

```python
answer = model.invoke(final_prompt)
print(answer.content)
```
